En este trabajo práctico tuvimos como finalidad programar un generador de señales senoidales que nos permita realizar futuras simulaciones.
En primer lugar cree la funcion: mi_funcion_sen, que devuelve un vector "tt" con valores de tiempo y un vector "xx" con valores de señal en hz. Y que toma como parámetros: 
"vmax: Amplitud máxima de la senoidal, dc: valor medio , ff: frecuencia, ph: fase en radianes , nn: numero de muestras , fs: frecuencia de muestreo" 

```python
import numpy as np
import matplotlib.pyplot as plt

"vmax: Amplitud maxima , dc: valor medio , ff: frecuencia, ph: fase en radianes , nn: numero de muestras , fs: frecuencia de muestreo" 

def mi_funcion_sen(vmax, dc , ff , ph , nn , fs):
    
    ts = 1/fs
    tt = np.arange(0, nn, 1)*ts
    xx = dc + vmax * np.sin(2 * np.pi * ff * tt + ph)

    return tt, xx 
"tt: vector de tiempo , xx: vector de señal"
```

Además, le agregamos ruido al resultado con un snr de 20 para comparar como cambia la señal con este.
Los parámetros utilizados en la funcion son: vmax = 1volt , dc = 0volts , ff = 1hz , ph = 0rad , nn = 1000 , fs = 1000.


```python
tt, xx = mi_funcion_sen(1 , 0 , 1 , 0 , 1000 , 1000)
snr = 20
pr = 10 ** (-snr/10)
ruido = np.random.normal(0 , np.sqrt(pr) , len(tt))

plt.figure(1)
plt.plot(tt,ruido,label = "Ruido")
plt.plot(tt,xx + ruido, label = "Seno con ruido")
plt.plot(tt,xx,label = "Seno 1hz")
plt.legend()
plt.show()
```

En este gráfico podemos observar la senoidal con 1hz en verde, un considerable ruido en azul y luego la senoidal con el ruido en naranja

A continuación como bonus usamos otros parámetros con distintas frecuencias, la primera que utilizaremos será de 500 hz:

```python
"BONUS"
tt2, xx2 = mi_funcion_sen(1 , 0 , 500 , 0 , 1000 , 1000)
xx2 = np.round(xx2, 11)
plt.figure(2)
plt.plot(tt2,xx2,label = "Seno 500hz" )
plt.legend()
plt.show()
```

En este gráfico podemos observar una linea horizontal que se mantiene constantemente en 0, cosa que tiene mucho sentido teniendo en cuenta que el seno de 500 hz utilizado es exactamente la mitad de los 1000 hz de la frecuencia de muestreo. Esto se debe a que estamos en la frecuencia de nyquist, donde nos quedaría un seno que siempre da 0 como resultado.
Algo extra para aclarar, es que la línea: "xx2 = np.round(xx2, 11)" fue utilizada ya que el compilador me devolvía varios valores al azar al graficar esta senoidal, por lo que tuve que redondearla para que funcione el experimento.

La siguiente frecuencia mostrada es de 999 hz:

```python
tt3, xx3 = mi_funcion_sen(1 , 0 , 999 , 0 , 1000 , 1000)
plt.figure(3)
plt.plot(tt3,xx3,label = "Seno 999hz" )
plt.legend()
plt.show()
```

Podemos observar que esta senoidal es exactamente como la de 1hz pero invertida en fase.

La siguiente frecuencia mostrada es la de 1001 hz:

```python
tt4, xx4 = mi_funcion_sen(1 , 0 , 1001 , 0 , 1000 , 1000)
plt.figure(4)
plt.plot(tt4,xx4,label = "Seno 1001hz" )
plt.legend()
plt.show()
```

Podemos observar que esta senoidal es exactamente como la de 1hz. Esto se debe a que una señal que difiere un múltiplo entero de la frecuencia de muestreo, genera la misma secuencia.

La siguiente frecuencia es de 2001hz:

```python
tt5, xx5 = mi_funcion_sen(1 , 0 , 2001 , 0 , 1000 , 1000)
plt.figure(5)
plt.plot(tt5,xx5,label = "Seno 2001hz" )
plt.legend()
plt.show()
```

Podemos observar que esta senoidal se ve igual a la de 1hz y 1001 hz, esto se debe a lo explicado anteriormente, el mismo resultado ocurriria al sumarlo por n veces la frecuencia de muestreo.
