```python
import numpy as np
import matplotlib.pyplot as plt

"vmax: Amplitud maxima , dc: valor medio , ff: frecuencia, ph: fase en radianes , nn: numero de muestras , fs: frecuencia de muestro" 

def mi_funcion_sen(vmax, dc , ff , ph , nn , fs):
    
    ts = 1/fs
    tt = np.arange(0, nn, 1)*ts
    xx = dc + vmax * np.sin(2 * np.pi * ff * tt + ph)

    return tt, xx 
"tt: vector de tiempo , xx: vector de señal"

tt, xx = mi_funcion_sen(1 , 0 , 1 , 0 , 1000 , 1000)
"BONUS"
tt2, xx2 = mi_funcion_sen(1 , 0 , 500 , 0 , 1000 , 1000)
tt3, xx3 = mi_funcion_sen(1 , 0 , 999 , 0 , 1000 , 1000)
tt4, xx4 = mi_funcion_sen(1 , 0 , 1001 , 0 , 1000 , 1000)
tt5, xx5 = mi_funcion_sen(1 , 0 , 2001 , 0 , 1000 , 1000)

snr = 20
pr = 10 ** (-snr/10)
ruido = np.random.normal(0 , np.sqrt(pr) , len(tt))


plt.figure(1)
plt.plot(tt,ruido,label = "Ruido")
plt.plot(tt,xx + ruido, label = "Seno con ruido")
plt.plot(tt,xx,label = "Seno 1000hz")
plt.legend()
plt.show()
"BONUS"
plt.figure(2)
plt.plot(tt2,xx2,label = "Seno 500hz" )
plt.legend()
plt.show()

plt.figure(3)
plt.plot(tt3,xx3,label = "Seno 999hz" )
plt.legend()
plt.show()

plt.figure(4)
plt.plot(tt4,xx4,label = "Seno 1001hz" )
plt.legend()
plt.show()

plt.figure(5)
plt.plot(tt5,xx5,label = "Seno 2001hz" )
plt.legend()
plt.show()


from scipy import signal as sig

n0 = 300
dd = np.zeros(1000)
dd[n0] = 1.

yy = sig.convolve(xx , dd)

yy2 = 1/1000*sig.convolve(ruido, np.flip(ruido))

plt.figure(6)
plt.plot(yy)

plt.figure(7)
plt.plot(yy2) 

B = 3
Vfs = 3
qq = Vfs / 2**B

xxq = np.round(xx / qq)

plt.figure(8)
plt.plot(xxq)

```
