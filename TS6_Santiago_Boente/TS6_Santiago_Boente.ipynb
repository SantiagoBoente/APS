En este trabajo práctico tuvimos como finalidad resolver ejercicios de transformada Z para los siguientes sistemas:

a) $y(n)=x(n−3)+x(n−2)+x(n−1)+x(n)$
 

b) $y(n)=x(n−4)+x(n−3)+x(n−2)+x(n−1)+x(n)$
 

c)  $y(n)=x(n)−x(n−1)$
 

d)  $y(n)=x(n)−x(n−2)$

 
Para realizar esto, haremos:


1) Hallar $T(z)=\frac{Y(z)}{X(z)}$
.
2) Calcular su respuesta en frecuencia de módulo y fase.

3) Simular y validar la respuesta en frecuencia de todos los sistemas con Numpy.

<img src="a.jpeg" align="center" width="550" /> 

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal


# =========================
# Sistema a)
# =========================

b_a = [1, 1, 1, 1]

w = np.linspace(0, np.pi, 1000)

z = np.exp(1j*w)

T = 1 + z**(-1) + z**(-2) + z**(-3) 

modulo = np.abs(T)
fase = np.unwrap(np.angle(T))

plt.figure(figsize=(10, 6))

plt.subplot(2, 1, 1)
plt.plot(w, modulo)
plt.title("Sistema a) - Respuesta en módulo")
plt.xlabel("Frecuencia [Hz]")
plt.ylabel("|H(f)| [dB]")
plt.grid(True)

plt.subplot(2, 1, 2)
plt.plot(w, fase)
plt.title("Sistema a) - Respuesta en fase")
plt.xlabel("Frecuencia [Hz]")
plt.ylabel("Fase [rad]")
plt.grid(True)

plt.tight_layout()
plt.show()
```
