```python
import numpy as np
import matplotlib.pyplot as plt

"vmax: Amplitud maxima , dc: valor medio , ff: frecuencia, ph: fase en radianes , nn: numero de muestras , fs: frecuencia de muestro" 

def mi_funcion_sen(vmax, dc , ff , ph , nn , fs):
    
    ts = 1/fs
    tt = np.arange(0, nn, 1)*ts
    xx = dc + vmax * np.sin(2 * np.pi * ff * tt + ph)

    return tt, xx 
"tt: vector de tiempo , xx: vector de señal"

tt, xx = mi_funcion_sen(1 , 0 , 1 , 0 , 1000 , 1000)
tt2, xx2 = mi_funcion_sen(1 , 0 , 1 , 0 , 1000 , 500)
tt3, xx3 = mi_funcion_sen(1 , 0 , 1 , 0 , 1000 , 999)
tt4, xx4 = mi_funcion_sen(1 , 0 , 1 , 0 , 1000 , 1001)
tt5, xx5 = mi_funcion_sen(1 , 0 , 1 , 0 , 1000 , 2001)

snr = 20
pr = 10 ** (-snr/10)
ruido = np.random.normal(0 , pr , len(tt))


plt.figure(1)
plt.plot(tt,xx,label = "Seno" )
plt.plot(tt,ruido,label = "Ruido")
plt.plot(tt,xx + ruido, label = "Seno con ruido")
plt.legend()
plt.show()

plt.figure(2)
plt.plot(tt2,xx2,label = "Seno" )
plt.legend()
plt.show()

plt.figure(3)
plt.plot(tt3,xx3,label = "Seno" )
plt.legend()
plt.show()

plt.figure(4)
plt.plot(tt4,xx4,label = "Seno" )
plt.legend()
plt.show()

plt.figure(5)
plt.plot(tt5,xx5,label = "Seno" )
plt.legend()
plt.show()
```
